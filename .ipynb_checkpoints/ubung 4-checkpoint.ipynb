```python

stars = ["Sirius", "Vega", "Capella", "Rigel"]


def find_star(stars_list, name):
    for star in stars_list:
        if star.lower() == name.lower(): 
            return star
    return None

# 1. Example when the star is found
search_name_1 = "Sirius"
star = find_star(stars, search_name_1)

if star is None:
    print("star not found for:", search_name_1 )
else:
    print(search_name_1, "is found")

# 2. Example when the star is not found
search_name_2 = "Betelgeuse"
star = find_star(stars, search_name_2)

if star is None:
    print("star not found for:", search_name_2 )
else:
     print(search_name_2, "is found")
```

```python
giant_stars = ["Pollux", "Betelgeuse", "Rigel"]

def find_star1(stars_list, name):

    if name in stars_list:
        return name
        
    for star in stars_list:
        # or is for flexible partial matching if the name is in current element of stars_list 
        if (star.lower() == name.lower() or name.lower() in star.lower()) and not (name == ""):
            return star
    return None

# 1. Example when the star is found
search_name_1 = "Sirius"
star = find_star1(giant_stars, search_name_1)

if star is None:
    print("star not found for:", search_name_1 )
else:
    print(search_name_1, "is found")

# 2. Example when the star is not found
search_name_2 = "Betelgeuse"
star = find_star1(giant_stars, search_name_2)

if star is None:
    print("star not found for:", search_name_2 )
else:
     print(search_name_2, "is found")

# 3. Example when the star is found in patial match
search_name_3 = "Betel"
star = find_star1(giant_stars, search_name_2)

if star is None:
    print("star not found for:", search_name_3 )
else:
     print(search_name_3, "is found")
```
