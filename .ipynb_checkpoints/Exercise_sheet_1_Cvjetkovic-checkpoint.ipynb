## Exercise 2 - Leap years

**(a)** Write a function that, given a year, reports if it is a [leap year](https://en.wikipedia.org/wiki/Leap_year). The tricky thing is that a leap year in the Gregorian calendar occurs:
 * on every year that is evenly divisible by 4
 * except every year that is evenly divisible by 100
 * unless the year is also evenly divisible by 400
 
For example, 1995 is not a leap year, but 1996 is. 1900 is not a leap year, but 2000 is.

**(b)** Test your function on some selected years that allow you to check if it provides the correct solution.

```python
def check_leap_year(year):
# Return "true" if year is a leap year, return "false" if it is not
    return (year%4 == 0 and year%100 != 0) or(year%400 == 0)

def main():
    try:
        #get the input value for the year
        value = input("Enter year to be checked:")
        year_to_check = int(value) # should be of integer type

        # check if the year is leap by calling the function
        if check_leap_year(year_to_check):
            print("entered year ",year_to_check, " is leap year")
        else:
            print("entered year ",year_to_check, " is NOT leap year")
            
    except ValueError:
        print("Invalid input. Please enter the year as an integer.")

if __name__ == "__main__":
    main()
```

## Exercise 3 - Rigged dice

**(a)** Create a rigged dice function that 25% of the time returns the number 6. The rest of the time it returns the integers 1,2,3,4,5 uniformly.

**(b)** Test your function, by calling it 100k times and checking that 6 is returned in the range of 24700-25300 (inclusive) times. You do not need to check that numbers 1 to 5 are returned uniformly or randomly, but you need to check that your function returns integers in the range 1-6 (inclusive). Such a test works roughly 98% of the time, so you might want to run it twice if you are confident your solution is correct.

```python
import random

# function to simulate one throw of the rigged dice, has no input parameters
def throw_rigged_dice():
    # 25% chance to have 6 thrown
    if random.random() < 0.25:
        return 6
    # 75% chance, choose 1, 2, 3, 4, or 5 to be thrown uniformly 
    return random.randint(1, 5)

def main():
    total_throws = 100000

    #using array to save the results of throwing the dice
    results = [0,0,0,0,0,0,0] 
    # index 0 is ignored, we have 6 possible values which will be saved in array within indice 1 to 7
    
    # Run the loop 100,000 times
    for i in range(1,total_throws,1):
        thrown_value = throw_rigged_dice() # one dice throw result is inside values 1,2,3,4,5,6
        results[thrown_value] = results[thrown_value] + 1
        
# report the final percentages on all dice throws
    print(f"Results after {total_throws:,} dice throws:")
# collect all the counts in array and calculate the percentages
    for thrown_value in range(1, 7):
        count = results[thrown_value]
        percentage = (count / total_throws) * 100
        print(f"Throw Result {thrown_value}: {percentage:.2f}%  count: {count:}")

if __name__ == "__main__":
    main()
```

## Exercise 4 - Find the fake coin

Consider this famous puzzle: _"You have 8 coins: all the coins have equal weight, except one. The odd one weights less than the others, not being of pure gold. In the worst case, how many iterations do you need to find the odd one out on a two plates scale?"_

Write a function that returns the minimum amount of weighing it will take to measure `N` coins in every possible occurrence (including worst case scenario, i.e.: without relying on luck at all!). The number `N` is guaranteed to be a positive integer.

**Tip:** being able to think recursively might help here!

**Note:** this is clearly a logical problem, because a good coder is a master of pattern recognition and subsequent optimization! ;)

```python
def process_weighing_steps(n, step=1):
#   Recursively calculates the minimum trials and shows steps 
#   on split of N coins - with the task to find the faked lighter coin.
    
    # Trivial case: if we have 1 coin then we require zero steps
    if n <= 1:
        return 0
    
    # Calculate 3 base group sizes fron "n" using integer division
    base_size = n // 3
    leftover = n % 3 #this can be 2 or 1 or 0
    
    # Distribute leftovers evenly on left and right plate: 
    # Left and Right plates must have an equal number of coins in the same weight to be in balance.
    if leftover == 2:
        group_a = base_size + 1  # Left plate
        group_b = base_size + 1  # Right plate
        group_c = base_size      # Off the scale
    elif leftover == 1:
        group_a = base_size      # Left plate
        group_b = base_size      # Right plate
        group_c = base_size + 1  # Off the scale
    else:
        group_a = base_size
        group_b = base_size
        group_c = base_size

    print(f"  Step {step}: Put {group_a} coins on Left Plate, {group_b} coins on Right Plate.")
    print(f"          Leave {group_c} coins off the scale.")
    
    # The worst-case remaining pool size of coins is the maximum of the three groups
    worst_case_group_size = max(group_a, group_b, group_c)
    print(f"          [Worst-case remaining pool for next step: {worst_case_group_size} coins]\n")
    
    return 1 + process_weighing_steps(worst_case_group_size, step + 1) #check next trial

if __name__ == "__main__":
#Test cases:
    print("###############################################")
    # Test case #1: The original 8 coins puzzle
    total_coins_1 = 8
    print(f"=== Instructions for {total_coins_1} Coins ===")
    total_trials_1 = process_weighing_steps(total_coins_1)
    print(f"Total trials needed in worst case: {total_trials_1}\n")
    print("###############################################")
    
    # Test case #2: A power of three (27 coins)
    total_coins_2 = 27
    print(f"=== Instructions for {total_coins_2} Coins ===")
    total_trials_2 = process_weighing_steps(total_coins_2)
    print(f"Total trials needed in worst case: {total_trials_2}\n")
    print("###############################################")

    # Test case #3: A power of three (81 coins)
    total_coins_3 = 81
    print(f"=== Instructions for {total_coins_3} Coins ===")
    total_trials_3 = process_weighing_steps(total_coins_3)
    print(f"Total trials needed in worst case: {total_trials_3}\n")
    print("###############################################")
```
