```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv('/home/vcvjet/exercise_01/exoplanet_data.csv')
df_filtered_1 = df[df["detection_type"] == "Imaging"]
df_filtered_2 = df[df["detection_type"] == "Astrometry"]
df_filtered_3 = df[df["detection_type"] == "Radial Velocity"]
df_filtered_4 = df[df["detection_type"] == "Primary Transit"]

plt.scatter(
	df_filtered_1["mass"],df_filtered_1["radius"],
	s=8,
	c='red',
	alpha = 0.5,
	label = 'Imaging'
#	edgecolors = 'black'
)

plt.scatter(
	df_filtered_2["mass"],df_filtered_2["radius"],
	s=8,
	c='green',
	alpha = 0.5,
	label = "Astrometry"
#	edgecolors = 'black'
)

plt.scatter(
	df_filtered_3["mass"],df_filtered_3["radius"],
	s=8,
	c='blue',
	alpha = 0.5,
	label = 'Radial Velocity'
#	edgecolors = 'black'
)

plt.scatter(
	df_filtered_4["mass"],df_filtered_4["radius"],
	s=8,
	c='orange',
	alpha = 0.5,
	label = 'Primary Transit'
#	edgecolors = 'black'
)



plt.xscale('log')
plt.yscale('log')
plt.xlabel('mass')
plt.ylabel('radius')
plt.title("planetary mass/radius")
plt.legend()
plt.show()
```
