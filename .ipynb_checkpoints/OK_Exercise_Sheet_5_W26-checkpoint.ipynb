## Exercise 1 - Numerical derivative

**(A)** Compute the derivative of the following two functions:

$$ f_1(x) = x^2 + \cos(x) $$

$$ f_2(x) = \exp(x) - x^3 $$

in correspondence of at least $N = 50$ values $x_i$ ($i = 1, ..., N$) of the variable $x$ in the interval $[1,5]$. To do this, *write your own code to compute the derivatives numerically*.

**(B)** Compare the resulting values you get for the numerical derivatives with the corresponding analytic solutions. Calculate the value of the quantity $q$ for each function:

$$ q = \sum_{i=1}^N \frac{\left[d_i - f'(x_i) \right]}{f'(x_i)} $$

where $d_i$ are the values you obtained for the numerical derivatives for each of the $N$ abscissas, and $f'(x_i)$ are the analytical derivatives evaluated in points $x_i$. How does the value of $q$ change when changing the value of $h$ to compute the derivatives numerically? How do the values of $q$ compare for the two functions? Discuss your findings.

````python
import numpy as np

# 1. Generate the domain grid and calculate step size (h)
N = 50
x = np.linspace(1, 5, N)
h = x[1] - x[0]  # Grid spacing

# 2. Define the two function arrays
y1 = x**2 + np.cos(x)
y2 = np.exp(x) - x**3

# 3. Compute derivatives manually using finite differences
def manual_derivative(y, h):
    # Initialize an array of zeros with the same length as y
    dy = np.zeros_like(y)
    
    # Interior points (index 1 to N-2): Central difference
    dy[1:-1] = (y[2:] - y[:-2]) / (2 * h)
    
    # Boundary points: Forward difference at start, Backward at end
    dy[0] = (y[1] - y[0]) / h
    dy[-1] = (y[-1] - y[-2]) / h
    
    return dy

# Execute the manual derivative calculation
df1_dx = manual_derivative(y1, h)
df2_dx = manual_derivative(y2, h)

# Print a subset of the results to inspect
print("x values (first 5):      ", x[:5])
print("f1 derivative (first 5): ", df1_dx[:5])
print("f2 derivative (first 5): ", df2_dx[:5])
````

````python
import numpy as np

# 1. Initialize grid boundaries and step size
N = 50
x = np.linspace(1, 5, N)
h = (5.0 - 1.0) / (N - 1)

# 2. Functional data arrays
y1 = x**2 + np.cos(x)
y2 = np.exp(x) - x**3

# 3. Compute true mathematical analytical derivatives
df1_analytical = 2 * x - np.sin(x)
df2_analytical = np.exp(x) - 3 * x**2

# 4. Manual differentiation logic 
def manual_derivative(y, h):
    dy = np.zeros_like(y)
    dy[1:-1] = (y[2:] - y[:-2]) / (2 * h)  # Central Difference (Interior)
    dy[0] = (y[1] - y[0]) / h              # Forward Difference (Start)
    dy[-1] = (y[-1] - y[-2]) / h           # Backward Difference (End)
    return dy

d1 = manual_derivative(y1, h)
d2 = manual_derivative(y2, h)

# 5. Calculate absolute relative total error metric (q)
q1 = np.sum(np.abs((d1 - df1_analytical) / df1_analytical))
q2 = np.sum(np.abs((d2 - df2_analytical) / df2_analytical))

print(f"Final q value for f1(x): {q1:.5f}")
print(f"Final q value for f2(x): {q2:.5f}")
````

---

### #1 Impact of Modifying the Step Size ($h$) on $q$

**As the grid step size $h$ decreases (by increasing $N$), the overall error metric $q$ drops sequentially.** 

However, because $q$ is defined as a *summation* over all $N$ grid coordinates ($\sum_{i=1}^N \text{Error}_i$), its scaling behavior reflects a blending of two competing numerical dynamics:
* **The Local Discretization Error:** At any individual interior coordinate, the central difference method has an error order of $\mathcal{O}(h^2)$. At the two boundary points, the one-sided differences have a lower accuracy order of $\mathcal{O}(h)$. As $h \to 0$, the mathematical error at every individual point gets smaller.
* **The Summation Accumulation:** When $h$ gets smaller, $N$ gets larger. Therefore, the metric $q$ has *more individual terms* to add together, which offsets some of the localized accuracy gains.

The net effect is that **$q$ scales approximately linearly with $h$** ($\mathcal{O}(h)$ overall). This happens primarily because the lower-accuracy boundary points dominate the total accumulation, and adding more interior terms bounds the structural convergence rate.

---

### #2 Comparison of $q$ Between the Two Functions

**The second function, $f_2(x) = e^x - x^3$, yields a significantly larger $q$ value than the first function, $f_1(x) = x^2 + \cos(x)$, across all step sizes.**

To see this clearly, look at how the metrics change across different configurations:

| Number of Points ($N$) | Step Size ($h$) | $q_1$ (for $f_1(x)$) | $q_2$ (for $f_2(x)$) | Error Ratio ($q_2 / q_1$) |
| :--- | :--- | :--- | :--- | :--- |
| $N = 25$ | $h \approx 0.1667$ | $0.14185$ | $1.59615$ | **~11.3x larger** |
| $N = 50$ | $h \approx 0.0816$ | $0.06898$ | $0.80767$ | **~11.7x larger** |
| $N = 100$ | $h \approx 0.0404$ | $0.03402$ | $0.42073$ | **~12.4x larger** |
| $N = 400$ | $h \approx 0.0100$ | $0.00842$ | $0.11250$ | **~13.4x larger** |

### Core Findings & Discussion

#### Finding A: The Impact of Higher-Order Derivatives
The remaining error in finite differences depends directly on the higher-order derivatives of the original function (via Taylor series expansion). 
* For $f_1(x)$, the derivatives remain small and bounded over $[1, 5]$ because the polynomials are low-degree ($x^2 \to 2 \to 0$) and the trigonometric term oscillates harmlessly ($\cos(x) \to -\sin(x) \to -\cos(x)$). 
* For $f_2(x)$, the exponential term $e^x$ is unique because its derivatives *never* vanish ($f^{(n)}(x) = e^x$). At the rightmost boundary ($x = 5$), $e^5 \approx 148.4$. This creates highly aggressive curves and rapid rates of change, resulting in much higher truncation errors for any fixed finite step size $h$.

#### Finding B: The Boundary Asymmetry Problem
The error in $f_2(x)$ is exacerbated near $x = 1.0$ because the analytical derivative crosses near zero ($f_2'(1) = e^1 - 3(1)^2 \approx -0.2817$). Because the formula for $q$ divides the individual error by the true analytical value ($f'(x_i)$), any absolute errors near zero are heavily amplified in the sum. $f_1(x)$, on the other hand, maintains clear, healthy values away from zero throughout $[1, 5]$, keeping its $q_1$ denominators stable.

## Exercise 3 - Gaussian Quadrature

**(A)** Compute the following integral by using a 4-point Gaussian quadrature. 

$$ I = \int_{-1}^{1} \cos(x) dx $$

To do this, use the Legendre polynomial of degree 4:

$$ P_4(x) = \frac{1}{8}(35 x^4 - 30 x^2 +3) $$

At this link: https://en.wikipedia.org/wiki/Gauss%E2%80%93Legendre_quadrature you can find the roots of Legendre polynomials and the necessary weights to solve the integral (i.e., you do not have to calculate them all from scratch!).

**(B)** Compare the result with the ones you obtain when using a Legendre polynomial of degree 3, 2, and 1, and comment your findings.

````python
import numpy as np

# 1. Define the Legendre polynomial coefficients for P_4(x) = (35x^4 - 30x^2 + 3) / 8
# The roots are invariant to the 1/8 scaling factor.
coefficients = [35, 0, -30, 0, 3]

# Compute the roots (quadrature nodes)
nodes = np.roots(coefficients) # finds all the roots (real and complex) of a polynomial where the equation equals zero.
nodes = np.sort(nodes)  # Sort nodes from smallest to largest

# 2. Define the derivative P_4'(x) = (140x^3 - 60x) / 8 = 0.5 * (35x^3 - 15x)
def p4_derivative(x):
    return 0.5 * (35 * x**3 - 15 * x)

# 3. Calculate the weights: w_i = 2 / ((1 - x_i^2) * [P_4'(x_i)]^2)
weights = 2 / ((1 - nodes**2) * (p4_derivative(nodes))**2)

# 4. Define the function to be integrated
def to_integrate_fun(x):
    return np.cos(x)

# 5. Compute the Gaussian Quadrature approximation: sum(w_i * f(x_i))
numerical_integral = np.sum(weights * to_integrate_fun(nodes))

# 6. Calculate the exact analytical value for comparison
exact_integral = 2 * np.sin(1)
error = abs(numerical_integral - exact_integral)

# Output results
print("4-Point Gaussian Quadrature Results:")
print("-" * 37)
for i in range(4):
    print(f"Node x_{i+1}: {nodes[i]:.8f} | Weight w_{i+1}: {weights[i]:.8f}")
print("-" * 37)
print(f"Numerical approximate calculation of the integral: {numerical_integral:.12f}")
print(f"Exact Analytical Value of the integral : {exact_integral:.12f}")
print(f"Absolute Error between Numerical and Analytical value: {error:.12e}")
````

````python
import numpy as np

# 1. Define the integrand function
def integrand(x):
    return np.cos(x)

# 2. Setup the exact analytical solution for comparison
exact_integral = 2 * np.sin(1)

# 3. Define the polynomial parameters for degrees 1 through 4
# The coefficients are ordered from highest power to lowest power.
# The respective derivatives P_n'(x) are used to compute the weights.
quadrature_rules = {
    1: {
        "name": "1-Point (Degree 1: P_1(x) = x)",
        "coeffs": [1, 0],
        "derivative": lambda x: np.ones_like(x)
    },
    2: {
        "name": "2-Point (Degree 2: P_2(x) = 0.5 * (3x^2 - 1))",
        "coeffs": [3, 0, -1],
        "derivative": lambda x: 3.0 * x
    },
    3: {
        "name": "3-Point (Degree 3: P_3(x) = 0.5 * (5x^3 - 3x))",
        "coeffs": [5, 0, -3, 0],
        "derivative": lambda x: 0.5 * (15 * x**2 - 3)
    },
    4: {
        "name": "4-Point (Degree 4: P_4(x) = 1/8 * (35x^4 - 30x^2 + 3))",
        "coeffs": [35, 0, -30, 0, 3],
        "derivative": lambda x: 0.5 * (35 * x**3 - 15 * x)
    }
}

# 4. Iterate through each rule and calculate the approximation
for degree, settings in quadrature_rules.items():
    # Compute and sort roots to get the quadrature nodes
    nodes = np.roots(settings["coeffs"])
    nodes = np.sort(nodes)
    
    # Calculate weights: w_i = 2 / ((1 - x_i^2) * [P_n'(x_i)]^2)
    p_prime = settings["derivative"](nodes)
    weights = 2 / ((1 - nodes**2) * (p_prime)**2)
    
    # Compute numerical integral approximation: sum(w_i * f(x_i))
    numerical_integral = np.sum(weights * integrand(nodes))
    error = abs(numerical_integral - exact_integral)
    
    # Print formatted output block matching the requested structure
    print(f"\n{settings['name']} Results:")
    print("-" * 46)
    for i in range(degree):
        print(f"Node x_{i+1}: {nodes[i]:.8f} | Weight w_{i+1}: {weights[i]:.8f}")
    print("-" * 46)
    print(f"Numerical Approximation : {numerical_integral:.12f}")
    print(f"Exact Analytical Value  : {exact_integral:.12f}")
    print(f"Absolute Error          : {error:.12e}")
````

````python
import numpy as np

# 1. Define the integrand function
def integrand(x):
    return np.cos(x)

# 2. Setup the exact analytical solution for comparison
exact_integral = 2 * np.sin(1)

# 3. Define standard derivative functions for each degree
def p1_derivative(x):
    return np.ones_like(x)

def p2_derivative(x):
    return 3.0 * x

def p3_derivative(x):
    return 0.5 * (15 * x**2 - 3)

def p4_derivative(x):
    return 0.5 * (35 * x**3 - 15 * x)


# ==========================================================
# DEGREE 1 CALCULATION
# ==========================================================
print("1-Point (Degree 1: P_1(x) = x) Results:")
print("-" * 46)

# FIXED: Added the correct coefficients [1, 0] for x + 0
coeffs_1 = [1, 0]
nodes_1 = np.sort(np.roots(coeffs_1))
p_prime_1 = p1_derivative(nodes_1)
weights_1 = 2 / ((1 - nodes_1**2) * (p_prime_1)**2)

numerical_1 = np.sum(weights_1 * integrand(nodes_1))
error_1 = abs(numerical_1 - exact_integral)

for i in range(1):
    print(f"Node x_{i+1}: {nodes_1[i]:.8f} | Weight w_{i+1}: {weights_1[i]:.8f}")
print("-" * 46)
print(f"Numerical Approximation : {numerical_1:.12f}")
print(f"Exact Analytical Value  : {exact_integral:.12f}")
print(f"Absolute Error          : {error_1:.12e}")


# ==========================================================
# DEGREE 2 CALCULATION
# ==========================================================
print("\n2-Point (Degree 2: P_2(x) = 0.5 * (3x^2 - 1)) Results:")
print("-" * 46)

coeffs_2 = [3, 0, -1]
nodes_2 = np.sort(np.roots(coeffs_2))
p_prime_2 = p2_derivative(nodes_2)
weights_2 = 2 / ((1 - nodes_2**2) * (p_prime_2)**2)

numerical_2 = np.sum(weights_2 * integrand(nodes_2))
error_2 = abs(numerical_2 - exact_integral)

for i in range(2):
    print(f"Node x_{i+1}: {nodes_2[i]:.8f} | Weight w_{i+1}: {weights_2[i]:.8f}")
print("-" * 46)
print(f"Numerical Approximation : {numerical_2:.12f}")
print(f"Exact Analytical Value  : {exact_integral:.12f}")
print(f"Absolute Error          : {error_2:.12e}")


# ==========================================================
# DEGREE 3 CALCULATION
# ==========================================================
print("\n3-Point (Degree 3: P_3(x) = 0.5 * (5x^3 - 3x)) Results:")
print("-" * 46)

coeffs_3 = [5, 0, -3, 0]
nodes_3 = np.sort(np.roots(coeffs_3))
p_prime_3 = p3_derivative(nodes_3)
weights_3 = 2 / ((1 - nodes_3**2) * (p_prime_3)**2)

numerical_3 = np.sum(weights_3 * integrand(nodes_3))
error_3 = abs(numerical_3 - exact_integral)

for i in range(3):
    print(f"Node x_{i+1}: {nodes_3[i]:.8f} | Weight w_{i+1}: {weights_3[i]:.8f}")
print("-" * 46)
print(f"Numerical Approximation : {numerical_3:.12f}")
print(f"Exact Analytical Value  : {exact_integral:.12f}")
print(f"Absolute Error          : {error_3:.12e}")


# ==========================================================
# DEGREE 4 CALCULATION
# ==========================================================
print("\n4-Point (Degree 4: P_4(x) = 1/8 * (35x^4 - 30x^2 + 3)) Results:")
print("-" * 46)

coeffs_4 = [35, 0, -30, 0, 3]
nodes_4 = np.sort(np.roots(coeffs_4))
p_prime_4 = p4_derivative(nodes_4)
weights_4 = 2 / ((1 - nodes_4**2) * (p_prime_4)**2)

numerical_4 = np.sum(weights_4 * integrand(nodes_4))
error_4 = abs(numerical_4 - exact_integral)

for i in range(4):
    print(f"Node x_{i+1}: {nodes_4[i]:.8f} | Weight w_{i+1}: {weights_4[i]:.8f}")
print("-" * 46)
print(f"Numerical Approximation : {numerical_4:.12f}")
print(f"Exact Analytical Value  : {exact_integral:.12f}")
print(f"Absolute Error          : {error_4:.12e}")

````

### Explanation of the Findings

* **Exponential Error Convergence:** As the polynomial degree ($n$) increases, the absolute error drops dramatically. Moving from a 1-point rule to a 4-point rule reduces the integration error by six orders of magnitude (from roughly $0.317$ down to $0.00000028$). This demonstrates the high efficiency of Gaussian quadrature for smooth, continuous transcendental functions like $\cos(x)$.
* **Algebraic Exactness:** An $n$-point Gaussian quadrature rule integrates any polynomial of degree $2n-1$ or less exactly. Because the Taylor series expansion of $\cos(x) = 1 - \frac{x^2}{2!} + \frac{x^4}{4!} - \dots$ converges very rapidly on the interval $[-1, 1]$, tracking higher-degree polynomial components instantly captures nearly all of the function's geometric area.
* **Symmetry of Nodes and Weights:** For every degree greater than 1, the calculated evaluation nodes ($x_i$) are perfectly symmetrical across the origin ($x=0$), and their corresponding weights ($w_i$) are identical. This symmetry stems from the even/odd nature of Legendre polynomials and mirrors the geometric symmetry of the $\cos(x)$ curve over $[-1, 1]$.
* **Conservation of Weight Mass:** Across all four distinct calculations, the sum of the weights ($\sum w_i$) always equals exactly $2.00000000$. This aligns with the physical definition of the system: if the integrand were a flat constant function $f(x)=1$ (a polynomial of degree 0), the total area under the curve over the interval $[-1, 1]$ must mathematically equal its width, which is $2$.
